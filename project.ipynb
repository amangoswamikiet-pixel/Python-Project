```python
# List of customer names
customer_names = ["Aman", "rahul", "sonam"]

# Each order is a tuple: (customer name, product, price, category)
order_details = [
    ("Aman", "Laptop", 999.99, "Electronics"),
    ("Aman", "Mouse", 24.99, "Electronics"),
    ("rahul", "Coffee Maker", 79.99, "Home Appliances"),
    ("sonam", "Notebook", 4.99, "Stationery"),
]

# Dictionary mapping each customer to their ordered products
customer_orders = {name: [] for name in customer_names}

for customer, product, price, category in order_details:
    customer_orders[customer].append(product)

print("Customers:", customer_names)
print("Order details:", order_details)
print("Customer orders:", customer_orders)
```

```python
from collections import Counter

# Illustrative customer and order data.
customer_names = ["Aman", "rahul", "sonam", "avik", "naveen", "aveer", "yuvan"]

# Each order is a tuple: (customer, product, price, category)
orders = [
    ("Aman", "Wireless earbuds", 89, "Electronics"),
    ("Aman", "Jeans", 45, "Clothing"),
    ("Aman", "Blender", 79, "Home Essentials"),
    ("rahul", "T-shirt", 25, "Clothing"),
    ("rahul", "Mug set", 18, "Home Essentials"),
    ("rahul", "Wireless earbuds", 89, "Electronics"),
    ("sonam", "Laptop", 899, "Electronics"),
    ("sonam", "Headphones", 149, "Electronics"),
    ("sonam", "Jacket", 120, "Clothing"),
    ("avik", "Towel set", 34, "Home Essentials"),
    ("avik", "Socks",12, "Clothing"),
    ("naveen", "Coffee maker", 74, "Home Essentials"),
    ("naveen", "Sweater", 59, "Clothing"),
    ("naveen", "Smartwatch", 199, "Electronics"),
    ("aveer", "T-shirt", 25, "Clothing"),
    ("aveer", "Blender", 79, "Home Essentials"),
    ("aveer", "Phone case", 19, "Electronics"),
    ("yuvan", "T-shirt", 25, "Clothing"),
    ("yuvan", "Storage basket", 32, "Home Essentials"),
]

print(orders)
```

```python
# map the code in dictionary for categories and customer
product_categories = {
    product: category
    for _, product, _, category in orders
}
categories = set(product_categories.values())
print(product_categories)
print(categories)
```

```python
customer_orders = {name: [] for name in customer_names}
print(customer_orders)
```

```python
customer_totals = {name: 0 for name in customer_names}
customer_categories = {name: set() for name in customer_names}
category_revenue = {category: 0 for category in categories}
product_frequency = Counter()

print(customer_totals)
print(customer_categories)
print(category_revenue)
print(product_frequency)
```

```python
for customer, product, price, category in orders:
    customer_orders.setdefault(customer, []).append(product)
    customer_totals[customer] += price
    customer_categories[customer].add(category)
    category_revenue[category] += price
    product_frequency[product] += 1

print(customer_orders)
print(customer_totals)
print(customer_categories)
print(category_revenue)
print(product_frequency)
```

```python
def classify_customer(total):
    if total > 100:
        return "High-value buyer"
    if total >= 50:
        return "Moderate buyer"
    return "Low-value buyer"


def money(amount):
    return f"${amount}"
```

```python
# write code analyze product and customer with unique category
unique_products = {product for _, product, _, _ in orders}
unique_categories = {category for _, _, _, category in orders}
print(unique_products)
print(unique_categories)


```

```python
# get the high value customer
high_value_customers = [
    name
    for name, total in customer_totals.items()
    if total > 100
]

print(high_value_customers)
```

```python
# get the most frequently purchased product

most_frequent_product = product_frequency.most_common(1)[0][0]
print(most_frequent_product)

```

```python
# Display the report.
print("CUSTOMER SPENDING SUMMARY")
for name, total in customer_totals.items():
    print(f"{name}: {money(total)} - {classify_customer(total)}")
```

```python
# total sales per category
print("\nCATEGORY SALES SUMMARY")
for category, total in category_revenue.items():
    print(f"{category}: {money(total)}")

```

```python
# key insights about purchase behavior
print("\nKEY INSIGHTS")
print(f"1. The most frequently purchased product is '{most_frequent_product}'.")
```

```python
# Identify trends based on category-wise sales
category_sales_trends = {
    category: sum(price for _, _, price, cat in orders if cat == category)
    for category in categories
}

print(category_sales_trends)
```
